Xavier initialization sets weights to maintain variance across layers,
improving convergence in deep network. Xavier Initialization, also known as Glorot Initialization, is used to initialize the weights of a neural network so that the variance of the activations is the same across every layer.

This helps prevent the vanishing or exploding gradient problem during training.s

```python
import numpy as np

def xavier_init(n_in, n_out, method='uniform'):
    if method == 'uniform':
        limit = np.sqrt(6 / (n_in + n_out))
        return np.random.uniform(-limit, limit, size=(n_in, n_out))
    elif method == 'normal':
        stddev = np.sqrt(2 / (n_in + n_out))
        return np.random.normal(0, stddev, size=(n_in, n_out))
    else:
        raise ValueError("Method must be 'uniform' or 'normal'")
# For a layer with 4 inputs and 3 outputs
weights = xavier_init(n_in=4, n_out=3, method='uniform')

print("Xavier Initialized Weights:\n", weights)
```

### Heaviside vs. Hard Sigmoid
Both Heaviside and Hard Sigmoid are activation functions, but they differ significantly in behavior and use cases.
- The Heaviside step function is a binary threshold function:

```python
import numpy as np

def heaviside(x):
    return np.where(x >= 0, 1.0, 0.0)
```

A Hard Sigmoid is a piecewise linear approximation of the sigmoid function:

```python
def hard_sigmoid(x):
    return np.clip(0.2 * x + 0.5, 0, 1)
```

### Swish Activation
Swish is a smooth, non-monotonic activation function introduced by researchers at Google. It has been found to perform better than ReLU in many deep learning tasks.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def swish(x):
    return x * sigmoid(x)
x = np.linspace(-5, 5, 100)
y = swish(x)

import matplotlib.pyplot as plt
plt.plot(x, y, label='Swish')
plt.title('Swish Activation Function')
plt.xlabel('x')
plt.ylabel('Swish(x)')
plt.grid(True)
plt.legend()
plt.show()
```
