Gradient Descent is an optimization algorithm used to minimize the loss function by updating the weights and biases of a neural network in the opposite direction of the gradient (slope) of the loss with respect to those parameters.

```python
import numpy as np

# Sigmoid activation and its derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training data (XOR example)
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

# Initialize weights and biases
np.random.seed(1)
input_size = 2
hidden_size = 4
output_size = 1

W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Training parameters
epochs = 10000
learning_rate = 0.1

# Training loop
for epoch in range(epochs):
    # ---- FORWARD PASS ----
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # ---- LOSS ----
    loss = np.mean((y - a2) ** 2)  # Mean squared error

    # ---- BACKPROPAGATION ----
    d_loss_a2 = 2 * (a2 - y)
    d_a2_z2 = sigmoid_derivative(a2)
    d_z2_W2 = a1
    d_loss_W2 = np.dot(d_z2_W2.T, d_loss_a2 * d_a2_z2)
    d_loss_b2 = np.sum(d_loss_a2 * d_a2_z2, axis=0, keepdims=True)

    d_z2_a1 = W2
    d_loss_a1 = np.dot(d_loss_a2 * d_a2_z2, d_z2_a1.T)
    d_a1_z1 = sigmoid_derivative(a1)
    d_loss_W1 = np.dot(X.T, d_loss_a1 * d_a1_z1)
    d_loss_b1 = np.sum(d_loss_a1 * d_a1_z1, axis=0, keepdims=True)

    # ---- GRADIENT DESCENT UPDATE ----
    W2 -= learning_rate * d_loss_W2
    b2 -= learning_rate * d_loss_b2
    W1 -= learning_rate * d_loss_W1
    b1 -= learning_rate * d_loss_b1

    # ---- PRINT LOSS ----
    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```
