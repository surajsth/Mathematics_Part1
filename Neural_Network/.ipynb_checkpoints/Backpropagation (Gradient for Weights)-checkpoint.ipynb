Backpropagation is the algorithm used in training neural networks to compute gradients of the loss with respect to each weight using the chain rule of calculus. These gradients are then used in gradient descent to update the weights.

```python
import numpy as np

# Sigmoid activation and derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)  # assuming x is already sigmoid(x)

# Inputs and true outputs
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])  # XOR

# Initialize weights and biases
np.random.seed(42)
W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

# Forward pass
z1 = X.dot(W1) + b1
a1 = sigmoid(z1)

z2 = a1.dot(W2) + b2
a2 = sigmoid(z2)

# Compute error
error = a2 - y
loss = np.mean(error ** 2)

# Backpropagation
d_a2 = error                # dLoss/da2
d_z2 = d_a2 * sigmoid_derivative(a2)
dW2 = a1.T.dot(d_z2)
db2 = np.sum(d_z2, axis=0, keepdims=True)

d_a1 = d_z2.dot(W2.T)
d_z1 = d_a1 * sigmoid_derivative(a1)
dW1 = X.T.dot(d_z1)
db1 = np.sum(d_z1, axis=0, keepdims=True)

# Print gradients
print("Gradient dW1:\n", dW1)
print("Gradient dW2:\n", dW2)
```
