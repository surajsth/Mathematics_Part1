Forward Propagation is the process of calculating the output of a neural network given an input. In a single-layer neural network, it's a simple linear transformation followed by an activation function.

```python
import numpy as np

# Activation function: Sigmoid (commonly used)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Input features (1 sample with 3 features)
X = np.array([0.5, 0.3, 0.2])  # shape: (3,)

# Initialize weights and bias
W = np.array([0.4, 0.7, 0.2])  # shape: (3,)
b = 0.1

# Forward propagation
z = np.dot(W, X) + b     # Linear transformation
a = sigmoid(z)           # Apply activation

print("Linear output (z):", z)
print("Activated output (a):", a)
```

 # Sigmoid Activation
The sigmoid activation maps inputs to [0, 1], commonly used for binary
classification.The Sigmoid activation function is a mathematical function used in neural networks to introduce non-linearity. It maps any real-valued input to the range (0, 1), which makes it ideal for binary classification problems.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Example usage
x = np.array([-2, -1, 0, 1, 2])
y = sigmoid(x)

print("Input:", x)
print("Sigmoid Output:", y)
```
