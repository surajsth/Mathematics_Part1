L2 Regularization, also called Weight Decay, adds a penalty term to the loss function to discourage large weights.
The idea is to:
- Penalize large weight values
- Prevent overfitting
- Encourage simpler models

```python
def l2_regularization_loss(weights, lambda_):
    return 0.5 * lambda_ * np.sum(weights ** 2)

def l2_gradient(weights, lambda_):
    return lambda_ * weights

def gradient_descent_with_l2(weights, grad_loss, lr, lambda_):
    # Total gradient = loss gradient + L2 gradient
    total_grad = grad_loss + l2_gradient(weights, lambda_)
    # Update weights
    weights -= lr * total_grad
    return weights
```

```python
import numpy as np

# Dummy values
weights = np.array([1.0, -2.0])
grad_loss = np.array([0.3, -0.5])
lambda_ = 0.1
lr = 0.01

# Update weights with L2 regularization
weights = gradient_descent_with_l2(weights, grad_loss, lr, lambda_)

print("Updated Weights:", weights)
```

### Maxout Activation
Maxout is an advanced activation function that generalizes ReLU and is especially useful in conjunction with dropout. Instead of applying a non-linear transformation (like sigmoid, tanh, or ReLU), Maxout selects the maximum from a set of linear functions.

```python
import numpy as np

class Maxout:
    def __init__(self, input_dim, output_dim, num_pieces):
        self.num_pieces = num_pieces
        self.output_dim = output_dim
        # Create k sets of weights and biases
        self.weights = np.random.randn(num_pieces, input_dim, output_dim)
        self.biases = np.random.randn(num_pieces, 1, output_dim)

    def forward(self, x):
        # Compute each linear piece
        z = np.array([x @ self.weights[i] + self.biases[i] for i in range(self.num_pieces)])
        # Take the element-wise maximum across all pieces
        return np.max(z, axis=0)
np.random.seed(0)
x = np.random.randn(3, 5)  # 3 samples, 5 features
maxout = Maxout(input_dim=5, output_dim=2, num_pieces=3)
output = maxout.forward(x)

print("Maxout Output:\n", output)
```
