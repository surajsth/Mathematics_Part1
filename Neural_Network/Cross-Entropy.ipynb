### Loss Function for Multi-Class (Cross-Entropy)
Cross-entropy loss measures the difference between two probability distributions:
- The true distribution (actual class labels)
- The predicted distribution (model's Softmax output)

```python
import numpy as np

def softmax(x):
    e_x = np.exp(x - np.max(x))  # for numerical stability
    return e_x / np.sum(e_x)

def cross_entropy_loss(y_true, y_pred):
    # Clip values to avoid log(0)
    y_pred = np.clip(y_pred, 1e-15, 1 - 1e-15)
    return -np.sum(y_true * np.log(y_pred))

# Example
logits = np.array([2.0, 1.0, 0.1])  # raw scores
y_true = np.array([1, 0, 0])        # true class is class 0

# Apply softmax to get probabilities
y_pred = softmax(logits)

# Calculate loss
loss = cross_entropy_loss(y_true, y_pred)

print("Predicted Probabilities:", y_pred)
print("Cross-Entropy Loss:", loss)
```
