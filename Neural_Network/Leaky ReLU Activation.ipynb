Leaky ReLU (Leaky Rectified Linear Unit) is a variant of the ReLU activation function designed to fix the "dying ReLU" problem. Unlike ReLU, which outputs 0 for all negative values, Leaky ReLU allows a small, non-zero gradient when the input is negative.

```python
import numpy as np

def leaky_relu(x, alpha=0.01):
    return np.where(x >= 0, x, alpha * x)

# Example usage
x = np.array([-3, -1, 0, 1, 3])
y = leaky_relu(x)

print("Input:", x)
print("Leaky ReLU Output:", y)
```

### ELU Activation (Exponential Linear Unit)
ELU (Exponential Linear Unit) is an activation function designed to fix some limitations of ReLU (like the "dying neuron" problem), while ensuring smoother and more flexible behavior, especially for negative values.

```python
import numpy as np

def elu(x, alpha=1.0):
    return np.where(x >= 0, x, alpha * (np.exp(x) - 1))

# Example usage
x = np.array([-3, -1, 0, 1, 3])
y = elu(x)

print("Input:", x)
print("ELU Output:", y)
```
