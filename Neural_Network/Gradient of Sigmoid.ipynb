The derivative of the sigmoid function is used in backpropagation to
compute gradients efficiently.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)
x = np.array([-2.0, 0.0, 2.0])
s = sigmoid(x)
ds = sigmoid_derivative(x)

print("Sigmoid Output:", s)
print("Sigmoid Derivative:", ds)
```

### RMSProp for Weight Updates
RMSProp adapts the learning rate for each weight based on the moving
average of squared gradients RMSProp (Root Mean Square Propagation) is an adaptive learning rate optimization algorithm designed to overcome issues with slow convergence and oscillations in gradient descent, particularly for non-stationary objectives.

It adjusts the learning rate independently for each parameter using a moving average of squared gradients..

```python
import numpy as np

def rmsprop_update(weights, gradients, cache, lr=0.01, decay_rate=0.9, epsilon=1e-8):
    # Update the running average of squared gradients
    cache = decay_rate * cache + (1 - decay_rate) * gradients**2
    
    # Update weights using RMSProp
    weights -= lr * gradients / (np.sqrt(cache) + epsilon)
    
    return weights, cache
# Initialize weights and gradients
weights = np.array([1.0, -2.0])
gradients = np.array([0.1, -0.3])
cache = np.zeros_like(weights)

# Perform update
weights, cache = rmsprop_update(weights, gradients, cache)

print("Updated Weights:", weights)
```
