### Tanh Activation
The Tanh (hyperbolic tangent) activation function is another popular function used in neural networks. It maps input values to the range (-1, 1) and is zero-centered, which often makes training faster than sigmoid.

```python
import numpy as np

def tanh(x):
    return np.tanh(x)

# Example usage
x = np.array([-2, -1, 0, 1, 2])
y = tanh(x)

print("Input:", x)
print("Tanh Output:", y)
```

### ReLU Activation
ReLU (Rectified Linear Unit) is the most commonly used activation function in modern neural networks. It introduces non-linearity by outputting the input directly if it's positive, otherwise 0. ReLU introduces non-linearity by zeroing negative values, often used
in deep networks

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

# Example usage
x = np.array([-2, -1, 0, 1, 2])
y = relu(x)

print("Input:", x)
print("ReLU Output:", y)
```

### Heaviside Step Activation
The Heaviside Step Function (also called the unit step function) is a threshold-based activation function used in early neural networks like the Perceptron. It outputs:
- 0 if the input is less than 0
- 1 if the input is greater than or equal to 0.
  
It’s a non-differentiable and non-continuous function, which makes it unsuitable for modern deep learning but conceptually important for understanding binary classification.

```python
import numpy as np

def heaviside_step(x):
    return np.where(x >= 0, 1, 0)

# Example usage
x = np.array([-2, -1, 0, 1, 2])
y = heaviside_step(x)

print("Input:", x)
print("Heaviside Step Output:", y)
```
