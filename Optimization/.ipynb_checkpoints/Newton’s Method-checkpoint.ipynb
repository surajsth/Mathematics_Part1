 Newton’s Method, one of the most powerful second-order optimization techniques out there! It uses curvature information to converge super fast, especially near the optimum.nlike gradient descent (which only uses the first derivative), Newton’s Method uses the second derivative (Hessian) to adjust the step size and direction.

```python
import numpy as np

# Example: f(x, y) = x^2 + 4y^2
def f(w):
    x, y = w
    return x**2 + 4*y**2

def grad(w):
    x, y = w
    return np.array([2*x, 8*y])

def hessian(w):
    return np.array([[2, 0],
                     [0, 8]])

# Initialization
w = np.array([2.0, 3.0])
for i in range(10):
    g = grad(w)
    H = hessian(w)
    delta = np.linalg.solve(H, g)  # H⁻¹ * g
    w = w - delta
    print(f"Iter {i}: w = {w}, f(w) = {f(w):.4f}")

print(f"\nMinimum found at: {w}")
```

# Proximal Gradient Method
Proximal Gradient Method — a super cool and elegant optimization method especially powerful for nonsmooth regularized problems like L1 norm (used in Lasso), group sparsity, and more.

```python
import numpy as np

def soft_thresholding(x, lam):
    return np.sign(x) * np.maximum(np.abs(x) - lam, 0)

# Simulated data
np.random.seed(0)
A = np.random.randn(100, 10)
x_true = np.random.randn(10)
x_true[5:] = 0  # make it sparse
b = A @ x_true + 0.1 * np.random.randn(100)

# PGD setup
x = np.zeros(10)
lr = 1e-2
lam = 0.1

for i in range(100):
    grad = A.T @ (A @ x - b)
    x = soft_thresholding(x - lr * grad, lr * lam)

    if i % 10 == 0:
        loss = 0.5 * np.linalg.norm(A @ x - b)**2 + lam * np.linalg.norm(x, 1)
        print(f"Iter {i}: loss = {loss:.4f}")

print(f"\nRecovered x:\n{x}")
```

# Proximal Gradient with L1 (ISTA)
ISTA (Iterative Shrinkage-Thresholding Algorithm) — a special case of the Proximal Gradient Method tailored specifically for L1 regularization. It's like the go-to algorithm for solving Lasso problems!

```python
import numpy as np

def soft_thresholding(x, thresh):
    return np.sign(x) * np.maximum(np.abs(x) - thresh, 0)

# Create synthetic data
np.random.seed(42)
n_samples, n_features = 100, 20
A = np.random.randn(n_samples, n_features)
x_true = np.random.randn(n_features)
x_true[10:] = 0  # sparsify
b = A @ x_true + 0.05 * np.random.randn(n_samples)

# ISTA parameters
x = np.zeros(n_features)
lambda_ = 0.1
lr = 1.0 / np.linalg.norm(A, ord=2)**2  # conservative learning rate
n_iters = 100

# ISTA loop
for i in range(n_iters):
    grad = A.T @ (A @ x - b)
    x = soft_thresholding(x - lr * grad, lr * lambda_)

    if i % 10 == 0:
        loss = 0.5 * np.linalg.norm(A @ x - b)**2 + lambda_ * np.linalg.norm(x, 1)
        print(f"Iter {i:3d}: loss = {loss:.4f}")

print("\nRecovered sparse vector:")
print(x)
```
