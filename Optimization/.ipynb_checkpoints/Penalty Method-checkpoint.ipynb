Penalty Method, a classic and foundational technique for handling constrained optimization problems by transforming them into unconstrained ones using penalty terms.

```python
import numpy as np
import matplotlib.pyplot as plt

# Original function
def f(x):
    return (x - 2)**2

# Penalty function for constraint x >= 1
def penalty(x):
    return np.maximum(0, 1 - x)**2

# Combined penalized objective
def penalized_f(x, mu):
    return f(x) + (1/mu) * penalty(x)

# Plot for different mu values
x_vals = np.linspace(-1, 3, 500)
for mu in [1, 0.1, 0.01]:
    plt.plot(x_vals, [penalized_f(x, mu) for x in x_vals], label=f"μ = {mu}")

plt.axvline(1, color='red', linestyle='--', label='Constraint x ≥ 1')
plt.title("Penalty Method: Effect of μ")
plt.xlabel("x")
plt.ylabel("Penalized Objective")
plt.legend()
plt.grid(True)
plt.show()
```

# Augmented Lagrangian Method
Augmented Lagrangian Method (ALM) — one of the most powerful and practical ways to solve constrained optimization problems, especially when traditional penalty methods get unstable. It's like a mix of penalty methods and Lagrange multipliers, and it fixes the weaknesses of both.

```python
import numpy as np
import matplotlib.pyplot as plt

# Objective and constraint
def f(x): return (x - 3)**2
def h(x): return x - 1

# Augmented Lagrangian
def L_aug(x, lam, mu):
    return f(x) + lam * h(x) + (mu / 2) * h(x)**2

# Initialization
x = 0.0
lam = 0.0
mu = 10.0

print("Iter |    x     |  f(x)  | Constraint | Lagrange Multiplier")
for i in range(10):
    # Minimize L_aug w.r.t x (simple case — 1D closed-form)
    x = (3 + lam + mu) / (1 + mu)
    
    # Update Lagrange multiplier
    lam = lam + mu * h(x)

    print(f"{i:4d} | {x:8.4f} | {f(x):6.4f} | {h(x):9.4f} | {lam:8.4f}")
```

# Dual Ascent Method
Dual Ascent Method — a classic technique in convex optimization that's especially powerful when dealing with constrained problems and distributed systems (like in network optimization, signal processing, or machine learning)

```python
import numpy as np

# Objective function and constraint
def f(x): return (x - 2)**2
def h(x): return x - 1
def lagrangian(x, lam): return f(x) + lam * h(x)

# Dual ascent
x = 0.0
lam = 0.0
lr = 0.1

print("Iter |   x   |  f(x) | h(x) | lambda")

for i in range(20):
    # Step 1: Primal update (minimize Lagrangian)
    x = (2 - lam) / 1  # since L(x, λ) = (x - 2)^2 + λ(x - 1), minimize gives x = 2 - λ

    # Step 2: Dual ascent
    lam += lr * h(x)

    print(f"{i:4d} | {x:5.3f} | {f(x):5.3f} | {h(x):5.3f} | {lam:5.3f}")
```
