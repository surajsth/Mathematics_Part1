Momentum method (aka the Heavy Ball method) is like giving your optimizer a little physics-based push so it moves faster and smoother toward the minimum.The momentum update introduces a velocity term that helps the optimizer build up speed in relevant directions and dampens oscillations in noisy or ravine-like loss surfaces.

```python
import numpy as np

# Function: f(x, y) = x^2 + y^2 -> min at (0, 0)
def f(w):
    return w[0]**2 + w[1]**2

def grad(w):
    return np.array([2*w[0], 2*w[1]])

# Initialize weights
w = np.array([5.0, -3.0])
v = np.zeros_like(w)
lr = 0.1
beta = 0.9

# Training loop
for i in range(50):
    g = grad(w)
    v = beta * v - lr * g
    w += v

    if i % 10 == 0:
        print(f"Iter {i}: w = {w}, f(w) = {f(w):.4f}")

print(f"\nFinal weights: {w}")
```

# Projection / Projected Gradient Descent
Projected Gradient Descent (PGD) — a powerful technique used when you're optimizing under constraints (like bounded variables or feasible regions). It’s commonly used in constrained optimization, adversarial attacks, and sparse learning.
PGD = Gradient Descent + Projection Step

```python
import numpy as np

# Simple quadratic function
def f(w):
    return np.sum((w - 0.5)**2)

def grad(w):
    return 2 * (w - 0.5)

# Projection onto [0, 1] box
def project_box(w):
    return np.clip(w, 0, 1)

# Init
w = np.array([1.5, -0.5])
lr = 0.1

for i in range(20):
    w -= lr * grad(w)
    w = project_box(w)  # Projection step

    if i % 5 == 0:
        print(f"Iter {i}: w = {w}, f(w) = {f(w):.4f}")

print(f"\nFinal projected w: {w}")
```
