# Quantile Regression Loss (Pinball Loss)
Quantile Regression models the conditional quantile (e.g., median, 90th percentile) of the response variable, rather than the mean. It is especially useful when the data is heteroscedastic (variance changes with inputs) or when you're interested in prediction intervals or asymmetric error penalties.

```python
import statsmodels.api as sm
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [2.3, 2.9, 3.7, 5.1, 8.2]
})

X = sm.add_constant(df['x'])
y = df['y']

# Fit quantile regression for the 90th percentile
quantile = 0.9
model = sm.QuantReg(y, X)
res = model.fit(q=quantile)

print(res.summary())
```
