# Gamma Regression Objective
Gamma regression models positive continuous data with a Gamma
distribution, often for skewed datasets Gamma Regression is a type of Generalized Linear Model (GLM) used for modeling continuous, positive, and skewed response variables — such as insurance claims, waiting times, or income — where the variance increases with the square of the mean.
.

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf
import pandas as pd

# Sample data
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [2.2, 2.8, 4.5, 7.1, 10.2]  # Positive continuous target
})

# Fit Gamma regression
model = smf.glm(formula='y ~ x', data=df, family=sm.families.Gamma(sm.families.links.log()))
result = model.fit()

print(result.summary())
```

# Probit Regression Model
Probit regression models binary classification using the cumulative
normal distribution function . Probit Regression is a type of Generalized Linear Model (GLM) used for modeling binary outcome variables — where the dependent variable 𝑌∈{0,1}Y∈{0,1}. It is very similar to logistic regression but uses a different link function.Φ

```python
import statsmodels.api as sm
import pandas as pd

# Example data
df = pd.DataFrame({
    'x': [1.2, 2.1, 3.3, 4.4, 5.2],
    'y': [0, 0, 1, 1, 1]
})

# Add constant for intercept
df['intercept'] = 1

# Fit Probit model
model = sm.Probit(df['y'], df[['intercept', 'x']])
result = model.fit()

print(result.summary())
```
