Lasso regression adds an L1 regularization term to the least squares
loss, promoting sparsity in the coefficients. Lasso = Least Absolute Shrinkage and Selection Operator

It’s a type of linear regression that uses L1 regularization to penalize large coefficients and force some coefficients to exactly zero → feature selection happens naturally!

```python
from sklearn.linear_model import Lasso

# Some fake data
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=100, n_features=10, noise=10, random_state=42)

# Apply Lasso
lasso = Lasso(alpha=0.1)  # alpha = lambda
lasso.fit(X, y)

print("Lasso Coefficients:", lasso.coef_)
```

# Ridge Regression Objective
Ridge regression adds an L2 regularization term to reduce overfitting
by shrinking coefficient. Ridge Regression is a type of linear regression that uses L2 regularization to shrink the model’s coefficients.

It penalizes large weights, but unlike Lasso, it does not force coefficients exactly to zero — it keeps all features but shrinks them closer to zero.s

```python
from sklearn.linear_model import Ridge

# Some fake data
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=100, n_features=10, noise=10, random_state=42)

# Apply Ridge
ridge = Ridge(alpha=1.0)  # alpha = lambda
ridge.fit(X, y)

print("Ridge Coefficients:", ridge.coef_)
```
