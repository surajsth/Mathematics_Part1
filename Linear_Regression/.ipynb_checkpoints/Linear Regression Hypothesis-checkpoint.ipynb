In simple linear regression, we model the relationship between an independent variable 𝑥 and a dependent variable 𝑦.
- 


```python
import numpy as np
X = np.array([[1, 2], [3, 4]])
beta = np.array([2, 3])
y_pred = X @ beta
```

```python
print(y_pred)
```

# Ordinary Least Squares (OLS)
OLS is the most common method for estimating the parameters in a linear regression model.
It works by minimizing the sum of the squared differences between the actual values and the predicted values (i.e., minimizing the error).
- 
- 

```python
import numpy as np

# Sample data (with intercept column added)
X = np.array([[1, 2],   # 1 is for intercept
              [3, 4]])
y = np.array([5, 11])

# OLS solution
beta_hat = np.linalg.inv(X.T @ X) @ X.T @ y

print("Estimated coefficients (β):", beta_hat)
```

# Mean Squared Error (MSE)
MSE is a commonly used metric to evaluate how well a regression model predicts the target variable.
- 
- 

```python
import numpy as np
from sklearn.metrics import mean_squared_error

# Actual and predicted values
y_true = np.array([1,2,3])
y_pred = np.array([1.1, 1.9, 3.2])

# Method 1: Using sklearn
mse = mean_squared_error(y_true, y_pred)
print("MSE (using sklearn):", mse)

# Method 2: Manual calculation
mse_manual = np.mean((y_true - y_pred) ** 2)
print("MSE (manual):", mse_manual)
```
