Negative Binomial Regression is a type of generalized linear model (GLM) used for modeling count data where:
- The response variable is a non-negative integer (0, 1, 2, …)
- The variance > mean (overdispersion), unlike Poisson which assumes variance = mean.
- It's often used in fields like epidemiology, insurance, ecology, and text mining.

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf
import pandas as pd

# Sample synthetic data
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [1, 3, 4, 8, 12]
})

# Fit Negative Binomial Regression
model = smf.glm(formula='y ~ x', data=df, family=sm.families.NegativeBinomial())
result = model.fit()

print(result.summary())
```
