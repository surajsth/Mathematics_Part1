R², or the coefficient of determination, tells you how well your regression model explains the variance in the target variable.
- 

```python
import numpy as np
from sklearn.metrics import r2_score

# Actual and predicted values
y_true = np.array([1,2,3])
y_pred = np.array([1.1, 1.9, 3.2])

# Method 1: Using sklearn
r2 = r2_score(y_true, y_pred)
print("R² Score (using sklearn):", r2)

# Method 2: Manual calculation
ss_res = np.sum((y_true - y_pred)**2)
ss_tot = np.sum((y_true - np.mean(y_true))**2)
r2_manual = 1 - (ss_res / ss_tot)
print("R² Score (manual):", r2_manual)
```

# Adjusted R²
Adjusted R² is a modified version of R² that adjusts for the number of predictors in the model. It's especially useful in multiple linear regression, where adding more variables can falsely inflate R².
- 

```python
import numpy as np
from sklearn.metrics import r2_score

# Sample data
y_true = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 8])
n = len(y_true)  # number of samples
k = 1            # number of predictors (you can change this)

# R²
r2 = r2_score(y_true, y_pred)

# Adjusted R²
r2_adj = 1 - ((1 - r2) * (n - 1) / (n - k - 1))

print("R²:", r2)
print("Adjusted R²:", r2_adj)
```
