## Huber Loss: Robust Regression Loss Function
Huber Loss is a loss function used in robust regression, combining the best of Mean Squared Error (MSE) and Mean Absolute Error (MAE). It is quadratic for small errors and linear for large errors — making it more robust to outliers than MSE, and more stable than MAE. Huber loss combines MSE and MAE, being quadratic for small errors
and linear for large errors, robust to outliers.

```python
from sklearn.linear_model import HuberRegressor
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt

# Create dataset with outliers
X, y = make_regression(n_samples=100, n_features=1, noise=5.0, random_state=42)
y[::10] += 50  # add outliers

# Fit Huber Regressor
huber = HuberRegressor(epsilon=1.35)  # epsilon ≈ δ
huber.fit(X, y)

# Plot
plt.scatter(X, y, label='Data')
plt.plot(X, huber.predict(X), color='red', label='Huber Regression')
plt.legend()
plt.title('Huber Loss Regression')
plt.show()
```
