ERM is the principle of choosing a model (hypothesis) that minimizes the average loss (error) on the training data.

```python
import numpy as np

# Simulated data
np.random.seed(0)
X = np.random.rand(100, 1)
y = 3 * X.squeeze() + 2 + np.random.randn(100) * 0.5

# Closed-form solution (ERM for squared loss)
X_b = np.c_[np.ones((100, 1)), X]  # add bias term
theta_hat = np.linalg.inv(X_b.T @ X_b) @ (X_b.T @ y)

print("ERM Solution (theta):", theta_hat)
```

# Logistic Regression Hypothesis
Logistic regression predicts probabilities for binary classification using
the sigmoid function applied to a linear combination of input. s

```python
import numpy as np

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Sample data
theta = np.array([0.5, -0.25])
x = np.array([1, 3])  # Assume x0=1 for bias

# Hypothesis
z = np.dot(theta, x)
h = sigmoid(z)

print(f"Predicted probability: {h:.4f}")
```

# Binary Cross-Entropy Loss
Binary cross-entropy measures the dissimilarity between predicted
probabilities and true labels in binary classification. Binary Cross-Entropy (BCE) is the loss function commonly used for binary classification tasks — like Logistic Regression or binary Neural Networks.

```python
import numpy as np

# Binary Cross Entropy Loss
def binary_cross_entropy(y_true, y_pred):
    # Clip predictions to avoid log(0)
    eps = 1e-15
    y_pred = np.clip(y_pred, eps, 1 - eps)
    return - (y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

# Example
y_true = np.array([1, 0, 1, 1])
y_pred = np.array([0.9, 0.2, 0.8, 0.7])

loss = binary_cross_entropy(y_true, y_pred)
print("Loss for each example:", loss)
print("Average loss:", np.mean(loss))
```

# Cross-Entropy Loss (Multi-Class)
Cross-entropy loss generalizes to multi-class classification, comparing
one-hot-encoded true labels with predicted probabilities Cross-Entropy Loss for Multi-Class Classification — the most popular loss for classification tasks like image recognition, text classification, etc! When you have more than two classes (e.g., cat 🐱, dog 🐶, rabbit 🐰), you use the Cross-Entropy Loss to measure how well your model's predicted probability distribution matches the true distribution..

```python
import numpy as np

def cross_entropy_loss(y_true, y_pred):
    # y_true is one-hot encoded
    eps = 1e-15
    y_pred = np.clip(y_pred, eps, 1 - eps)
    return -np.sum(y_true * np.log(y_pred))

# Example
y_true = np.array([1, 0, 0])  # True class is "Cat"
y_pred = np.array([0.7, 0.2, 0.1])  # Model's prediction

loss = cross_entropy_loss(y_true, y_pred)
print(f"Cross-Entropy Loss: {loss:.4f}")
```

# Hinge Loss for SVM
Hinge loss penalizes predictions that are not at least 1 margin away
from the correct classification in support vector machines (SVMs)Hinge Loss is used for maximum-margin classification

```python
import numpy as np

def hinge_loss(y_true, y_pred):
    return np.maximum(0, 1 - y_true * y_pred)

# Example
y_true = np.array([1, -1, 1, -1])
y_pred = np.array([2.0, -0.5, 0.3, -2.0])  # raw scores from model

loss = hinge_loss(y_true, y_pred)
print("Hinge loss for each sample:", loss)
print("Average loss:", np.mean(loss))
```
