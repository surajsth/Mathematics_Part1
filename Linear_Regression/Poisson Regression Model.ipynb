Poisson Regression is a type of generalized linear model (GLM) used to model count data 
— where the dependent variable is a non-negative integer (0, 1, 2, …).

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf
import pandas as pd

# Sample data
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [0, 1, 2, 5, 8]  # count data
})

# Fit Poisson Regression Model
model = smf.glm(formula='y ~ x', data=df, family=sm.families.Poisson())
result = model.fit()

print(result.summary())
```
