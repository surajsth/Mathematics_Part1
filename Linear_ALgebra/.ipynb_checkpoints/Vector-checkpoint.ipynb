# Vector Addition

```python
import numpy as np
```

```python
a = np.array([1,2])
b = np.array([3,4])
```

```python
result = a+b
print(result)
```

# Vector multiplication

```python
result = a*b
print(result)
```

# scalar multiplcation

```python
alpha = 3
u = np.array([-1, 3])
```

```python
scalar_mul = alpha * u
print(scalar_mul)
```

# Dot Product

```python
result = np.dot(a, b)
print(result)
```

# Cross Product

```python
a = np.array([1,0,0])
b = np.array([0,1,0])
```

```python
result = np.cross(a, b)
print(result)
```

# Norm of a Vector (Euclidean)

∥v∥ =√(3^2 + 4^2) = 5

```python
v = np.array([3, 4])
```

```python
result = np.linalg.norm(v)
print(result)
```

# Orthogonality Condition

```python
u = np.array([1, 2])
v = np.array([-2, 1])
```

```python
result = np.dot(u, v)
print(result)
```

```python
is_orthogonal = result == 0
print(is_orthogonal)
```
