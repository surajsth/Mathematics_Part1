# Binary Cross-Entropy (special case)
Binary Cross-Entropy (BCE) is a loss function commonly used in binary classification tasks, where the output is a binary label (0 or 1). It measures the difference between the true label and the predicted probability in a binary classification setting. BCE is widely used in logistic regression, binary classification neural networks, and other similar models.
* Example: For y = [1, 0] and ˆy = [0.8, 0.2], compute BCE = −1/2(log(0.8) + log(0.8)).

```python
import numpy as np
y = np.array([1, 0])
y_hat = np.array([0.8, 0.2])
bce = -np.mean(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))
print(bce)
```

```python
# single batch
import numpy as np

# True label and predicted probability
y_true = 1
y_pred = 0.9  # predicted probability of class 1

# Calculate Binary Cross-Entropy
bce_loss = - (y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
print(f"Binary Cross-Entropy Loss: {bce_loss:.4f}")
```

```python
# multiple predictions
y_true = np.array([1, 0, 1, 1, 0])  # true labels
y_pred = np.array([0.9, 0.2, 0.8, 0.95, 0.1])  # predicted probabilities

# Calculate Binary Cross-Entropy for each sample
bce_loss = - (y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

# Calculate average BCE loss
average_bce_loss = np.mean(bce_loss)
print(f"Average Binary Cross-Entropy Loss: {average_bce_loss:.4f}")
```
