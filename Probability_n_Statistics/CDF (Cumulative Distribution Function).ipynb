# CDF (Cumulative Distribution Function)
* The CDF of a random variable gives the probability that the variable
takes a value less than or equal to x. It is used to describe the distribution function for
both discrete and continuous variables.

```python
from scipy.stats import uniform
x = 0.5
cdf_value = uniform.cdf(x, loc=0, scale=1)
print(cdf_value)
```

```python
# CDF for Discrete Distributions
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10   # Number of trials
p = 0.5  # Probability of success

# Generate possible values
x = np.arange(0, n+1)
cdf_values = binom.cdf(x, n, p)  # Compute CDF
print(cdf_values)

# Plot
plt.step(x, cdf_values, where='mid', label='Binomial CDF', color='blue')
plt.xlabel("Number of Successes (X)")
plt.ylabel("Cumulative Probability P(X ≤ x)")
plt.title("CDF of Binomial Distribution (n=10, p=0.5)")
plt.legend()
plt.grid()
plt.show()
```

```python
# CDF for Continuous Distributions
from scipy.stats import norm

# Parameters
mu = 0     # Mean
sigma = 1  # Standard deviation

# Generate x values
x = np.linspace(-4, 4, 1000)

# Compute CDF
cdf_values = norm.cdf(x, mu, sigma)

# Plot
plt.plot(x, cdf_values, 'r-', label="Normal CDF")
plt.xlabel("x")
plt.ylabel("Cumulative Probability P(X ≤ x)")
plt.title("CDF of Normal Distribution (μ=0, σ=1)")
plt.legend()
plt.grid()
plt.show()
```
