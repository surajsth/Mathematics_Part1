# Joint Probability
* Link https://www.investopedia.com/terms/j/jointprobability.asp
* Joint probability quantifies the likelihood of two events occurring
together.
* P(A ∩ B) = P(A | B)P(B)

```python
# Joint Probability for Discrete Variables (PMF)
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Define two discrete random variables (X, Y)
n = 5   # Number of trials
p_x = 0.5  # Probability of success for X
p_y = 0.7  # Probability of success for Y

# Generate possible values
x_vals = np.arange(0, n+1)
y_vals = np.arange(0, n+1)

# Compute joint probabilities (assuming independence)
joint_pmf = np.outer(binom.pmf(x_vals, n, p_x), binom.pmf(y_vals, n, p_y))

# Print joint probability table
print("Joint PMF Table:")
print(joint_pmf)
```

```python
# Joint Probability for Continuous Variables (PDF)
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal

# Define mean and covariance matrix for two normal variables
mu = [0, 0]  # Mean vector (mu_x, mu_y)
cov = [[1, 0.5], [0.5, 1]]  # Covariance matrix

# Create grid of (x, y) values
x, y = np.meshgrid(np.linspace(-3, 3, 100), np.linspace(-3, 3, 100))
pos = np.dstack((x, y))

# Compute joint PDF
rv = multivariate_normal(mu, cov)
pdf_vals = rv.pdf(pos)

# Plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, pdf_vals, cmap="viridis")

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Joint PDF')
ax.set_title('Joint PDF of Two Normal Variables')

plt.show()
```

```python
 # Computing Joint Probability from a Dataset
import pandas as pd

# Sample dataset (discrete variables)
data = {'X': ['A', 'A', 'B', 'A', 'B', 'B', 'B', 'A', 'B', 'A'],
        'Y': ['C', 'D', 'C', 'D', 'C', 'C', 'D', 'C', 'C', 'D']}

df = pd.DataFrame(data)

# Compute joint probability table
joint_prob = df.value_counts(normalize=True).unstack()

print("Joint Probability Table:")
print(joint_prob)
```
