# Entropy (discrete)
Entropy measures the uncertainty of a discrete random variable. It is
a fundamental concept in information theory and ML, particularly in decision trees and
loss function
* H(X)=− i∑ P(x i)log b P(x i).

```python
import numpy as np

def entropy(prob_dist, base=2):
    prob_dist = np.array(prob_dist)
    prob_dist = prob_dist[prob_dist > 0]  # Avoid log(0)
    return -np.sum(prob_dist * np.log(prob_dist) / np.log(base))

# Example discrete probability distribution
p = [0.5, 0.25, 0.25]  # Must sum to 1
H = entropy(p)
print(f"Entropy: {H:.4f} bits")
```

```python
from scipy.stats import entropy

p = [0.5, 0.25, 0.25]
H = entropy(p, base=2)
print(f"Entropy: {H:.4f} bits")
```

```python
# Fair 6-sided die
p = [1/6] * 6
H = entropy(p, base=2)
print(f"Entropy of fair die: {H:.4f} bits")  # Should be close to log2(6)
```
