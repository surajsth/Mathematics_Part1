# Conditional Expectation
Conditional expectation is the expected value of a random variable X
given that another variable Y is known. It is critical in Bayesian inference and probabilistic modeling.* Example: If X = {1, 2} with P(X = 1 | Y ) = 0.7 and P(X = 2 | Y ) = 0.3, then
E[X | Y ] = 1 · 0.7 + 2 · 0.3 = 1.3.



```python
X = [1, 2]
P_X_given_Y = [0.7, 0.3]
conditional_expectation = sum(x * p for x, p in zip(X, P_X_given_Y))
print(conditional_expectation)
```

```python
import numpy as np

# Define joint probability table P(X, Y)
joint_p = np.array([
    [0.1, 0.2],  # X = 1
    [0.2, 0.1],  # X = 2
    [0.1, 0.3],  # X = 3
])

x_vals = np.array([1, 2, 3])
y_vals = np.array([0, 1])

# Marginal P(Y)
p_y = joint_p.sum(axis=0)

# Conditional P(X | Y=y)
p_x_given_y = joint_p / p_y  # Broadcasting

# Conditional Expectation E[X | Y=y]
e_x_given_y = (x_vals[:, None] * p_x_given_y).sum(axis=0)

for i, y in enumerate(y_vals):
    print(f"E[X | Y={y}] = {e_x_given_y[i]:.4f}")
```

```python
import pandas as pd

# Example data
df = pd.DataFrame({
    'X': [1, 2, 3, 1, 2, 3, 1, 2, 3],
    'Y': [0, 0, 0, 1, 1, 1, 1, 1, 1],
    'P': [0.1, 0.2, 0.1, 0.2, 0.1, 0.3, 0, 0, 0]  # Joint probabilities
})

# Normalize to ensure probabilities sum to 1
df['P'] /= df['P'].sum()

# Compute conditional expectation E[X | Y=y]
result = df.groupby('Y').apply(lambda g: (g['X'] * g['P']).sum() / g['P'].sum())
print(result)
```
