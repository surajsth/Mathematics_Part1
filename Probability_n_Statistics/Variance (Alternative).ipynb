# Variance (Alternative)
An alternative formula for variance uses the difference between the
expected value of the square of X and the square of the expected value of X. ariance is a fundamental concept in probability and statistics that measures the spread or dispersion of a set of data points or a random variable around its mean. 

```python
import numpy as np
X = np.array([1, 2, 3])
E_X2 = np.mean(X**2)
E_X = np.mean(X)
variance = E_X2 - E_X**2
print("Variance ",variance)
```
