# Skewness
Skewness measures the asymmetry of the probability distribution of a
random variable about its mean. Positive skew indicates a longer right tail, and negative
skew indicates a longer left tl. --> * Skewness(X) = E[(X − µ)**3]σ* / 3

3

For X = {1, 2, 3} with mean µ = 2 and standard deviation σ = 0.816,
compute Skewness(X) using the third central moment.

```python
import numpy as np
X = [1, 2, 3]
mu = np.mean(X)
sigma = np.std(X)
skewness = np.mean(((X - mu) / sigma)**3)
print("Skewness", skewness)
```

```python
import numpy as np
from scipy.stats import skew

# Example dataset
data = [2, 4, 5, 5, 6, 6, 7, 8, 9, 10, 12, 15, 18]

# Calculate skewness
skew_value = skew(data)
print(f"Skewness: {skew_value:.4f}")
```

```python
import matplotlib.pyplot as plt

# Plotting the histogram of the data
plt.hist(data, bins=10, edgecolor='black')
plt.title(f"Histogram (Skewness: {skew_value:.2f})")
plt.xlabel('Data Values')
plt.ylabel('Frequency')
plt.show()
```
