# Probability Mass Function (PMF)¶
* The PMF defines the probabilities of discrete outcomes of a random variable

* PY(k)=P(Y=k)=(1−p)*k−1p, for k=1,2,3,...
* Example: If X = {1, 2, 3} with P(X = 1) = 0.2, P(X = 2) = 0.5, and P(X = 3) = 0.3, the PMF is defined for these values.

```python
# Define X values and their probabilities
X = [1, 2, 3]
P_X = [0.2, 0.5, 0.3]

# Define PMF function
def pmf(x):
    return P_X[X.index(x)] if x in X else 0  # Return P(X=x) if x exists, else return 0

# Test the PMF function
print(f"P(X=1): {pmf(1)}")  # Expected Output: 0.2
print(f"P(X=2): {pmf(2)}")  # Expected Output: 0.5
print(f"P(X=3): {pmf(3)}")  # Expected Output: 0.3
print(f"P(X=4): {pmf(4)}")  # Expected Output: 0 (since 4 is not in X)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10   # number of trials
p = 0.5  # probability of success
x = np.arange(0, n+1)  # Possible values from 0 to n
pmf_values = binom.pmf(x, n, p)

print(f"P(X={x}) = {pmf_values}")

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(x, pmf_values, 'bo-', markersize=8, label='PMF')  # 'bo-' makes blue dots with a line
plt.vlines(x, 0, pmf_values, colors='blue', lw=2, alpha=0.5)  # Vertical lines for clarity

# Labels and title
plt.xlabel("Number of Successes (k)")
plt.ylabel("Probability P(X = k)")
plt.title("Binomial Distribution PMF (n=10, p=0.5)")
plt.legend()
plt.grid()
plt.show()
```
