# Conditional Probability
Example: If P(A ∩ B) = 0.2 and P(B) = 0.5, then P(A | B) = 0.2/
0.5 = 0.4.

```python
P_B = 0.5  # Probability of event B
P_A_intersect_B = 0.2

P_A_given_B_computed = P_A_intersect_B / P_B  # P(A | B)

print(f"P(A | B): {P_A_given_B_computed}")
```

# Law of Total Probability
If P(A | B1) = 0.3, P(A | B2) = 0.7, P(B1) = 0.4, and P(B2) = 0.6, then
P(A) = 0.3*· 0.4 + 0.7*· 0.6 = 0.54.

```python
P_A_intersect_B1 = 0.3
P_A_intersect_B2 = 0.7
P_B1 = 0.4
P_B2 = 0.6

P_A = P_A_intersect_B1 * P_B1 + P_A_intersect_B2 * P_B2  

print(f"P(A): {P_A}")
```

# Bayes’ Theorem
Example: If P(B | A) = 0.8, P(A) = 0.3, and P(B) = 0.5, then P(A | B) = (0.8-0.3)/
0.5 
0.48.

```python
P_B_intersect_A = 0.8
P_A = 0.3
P_B = 0.5

P_A_intersect_B = (P_B_intersect_A - P_A) / P_B
print(f"P(A | B): {P_A_intersect_B}")
```

# Expectation* The expectation (mean) of a random variable is the weighted average
of all possible values, weighted by their probabilities
* Example: If X = {1, 2, 3} with P(X = 1) = 0.2, P(X = 2) = 0.5, and P(X = 3) =
0.3, then E[X] = 1 · 0.2 + 2 · 0.5 + 3 · 0.3 = 2.
  ..

```python
X = [1, 2, 3]
P_X = [0.2, 0.5, 0.3]

# Compute expected value
E_X = sum(x * p for x, p in zip(X, P_X))

print(f"Expected Value (E[X]): {E_X}")
```

# Variance
Example: For X = {1, 2, 3} with P(X = 1) = 0.2, P(X = 2) = 0.5, and P(X = 
3) = 0.3, compute E[X] = 2.1 and E[X^2
] = 9.7
, so Var(X) = 9.7 − (2. 1)2 = 4.29.

```python
# Define values of X and their probabilities
X = [1, 2, 3]
P_X = [0.2, 0.5, 0.3]

# Compute expectation (E[X])
expectation = sum(x * p for x, p in zip(X, P_X))

# Compute expectation of X^2 (E[X^2])
expectation_X2 = sum(x**2 * p for x, p in zip(X, P_X))

# Compute variance (Var(X))
variance = expectation_X2 - expectation**2

print(f"Expected Value (E[X]): {expectation}")
print(f"E[X^2]: {expectation_X2}")
print(f"Variance (Var(X)): {variance}")
```

#  Standard Deviation
If Var(X) = 0.29, then σ(X) = √
0.29 ≈ 0.54.

```python
variance = 0.29
std_dev = variance**0.5
print(f"Standard Deviation (σ(X)): {std_dev}")
```

# Covariance
Example: If X = {1, 2}, Y = {3, 4}, P(X, Y ) = {0.5, 0.5}, and E[X] = 1.5, E[Y ] =
3.5, compute Cov(X, Y ) = 0.25.

```python
X = [1, 2]
Y = [3, 4]
P_XY = [0.5, 0.5]
E_X = sum(x * p for x, p in zip(X, P_XY))
E_Y = sum(y * p for y, p in zip(Y, P_XY))
covariance = sum((x - E_X) * (y - E_Y) * p for x, y, p in zip(X, Y, P_XY))
print(f"covariance (COV(X, Y)): {covariance}")
```

# Correlation
Example: If Cov(X, Y ) = 0.25, σ(X) = 0.5, and σ(Y ) = 1.0, then ρ(X, Y ) =
0.25
0.5·1.0 = 0.5.

```python
covariance = 0.25
std_X = 0.5
std_Y = 1.0
correlation = covariance / (std_X * std_Y)
print(f"correlation (P(X, Y)): {correlation}")
```

# Probability Mass Function (PMF)
The PMF defines the probabilities of discrete outcomes of a random
variable
* PY(k)=P(Y=k)=(1−p)**k−1*p, for k=1,2,3,...
* Example: If X = {1, 2, 3} with P(X = 1) = 0.2, P(X = 2) = 0.5, and P(X = 3) =
0.3, the PMF is defined for these values.

```python
# Define X values and their probabilities
X = [1, 2, 3]
P_X = [0.2, 0.5, 0.3]

# Define PMF function
def pmf(x):
    return P_X[X.index(x)] if x in X else 0  # Return P(X=x) if x exists, else return 0

# Test the PMF function
print(f"P(X=1): {pmf(1)}")  # Expected Output: 0.2
print(f"P(X=2): {pmf(2)}")  # Expected Output: 0.5
print(f"P(X=3): {pmf(3)}")  # Expected Output: 0.3
print(f"P(X=4): {pmf(4)}")  # Expected Output: 0 (since 4 is not in X)
```
