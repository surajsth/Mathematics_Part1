# Kurtosis
Kurtosis measures the tailedness of a distribution and is used to detect outliers.Kurtosis measures the ”tailedness” of the probability distribution. A
high kurtosis indicates heavy tails, while a low kurtosis indicates light tails
##### 
Example: For X = {1, 2, 3} with mean µ = 2 and standard deviation σ = 0.816,
compute Kurtosis(X) using the fourth central moment.

```python
import numpy as np
X = [1, 2, 3]
mu = np.mean(X)
sigma = np.std(X)
kurtosis = np.mean(((X - mu) / sigma)**4)
print(kurtosis)
```

```python
import numpy as np
from scipy.stats import kurtosis

# Example dataset
data = [2, 4, 5, 5, 6, 6, 7, 8, 9, 10, 12, 15, 18]

# Calculate kurtosis
kurt_value = kurtosis(data)
print(f"Kurtosis: {kurt_value:.4f}")
```

```python
import matplotlib.pyplot as plt

# Plotting the histogram of the data
plt.hist(data, bins=10, edgecolor='black')
plt.title(f"Histogram (Kurtosis: {kurt_value:.2f})")
plt.xlabel('Data Values')
plt.ylabel('Frequency')
plt.show()
```
