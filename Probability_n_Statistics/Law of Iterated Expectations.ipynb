# Law of Iterated Expectations
The law of iterated expectations states that the expectation of X is the
weighted average of its conditional expectations over Y . It is foundational in probability
theory and statistics.
* E[X] = E[E[X | Y 
* Example: Suppose X depends on Y = {1, 2}, with E[X | Y = 1] = 3, E[X | Y =
2] = 5, and P(Y = 1) = 0.6, P(Y = 2) = 0.4. Then E[X] = 3 · 0.6 + 5 · 0.4 = 3.8]]

```python
E_X_given_Y = [3, 5]
P_Y = [0.6, 0.4]
E_X = sum(e * p for e, p in zip(E_X_given_Y, P_Y))
print(E_X)
```

```python
import numpy as np

# Joint distribution P(X, Y)
joint_p = np.array([
    [0.1, 0.2],  # X=1
    [0.2, 0.1],  # X=2
    [0.1, 0.3],  # X=3
])

x_vals = np.array([1, 2, 3])
y_vals = np.array([0, 1])

# Step 1: Marginal P(Y)
p_y = joint_p.sum(axis=0)

# Step 2: Conditional P(X|Y)
p_x_given_y = joint_p / p_y  # shape (3, 2)

# Step 3: E[X | Y=y] for each y
e_x_given_y = (x_vals[:, None] * p_x_given_y).sum(axis=0)  # shape (2,)

# Step 4: E[E[X | Y]]
e_x = (e_x_given_y * p_y).sum()

print(f"E[X | Y=0] = {e_x_given_y[0]:.4f}")
print(f"E[X | Y=1] = {e_x_given_y[1]:.4f}")
print(f"E[X] = {e_x:.4f}")
```
