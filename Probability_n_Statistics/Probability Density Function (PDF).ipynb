# Probability Density Function (PDF)
* The PDF defines the relative likelihood of a continuous random variable at a specific value.

* For more https://www.investopedia.com/terms/p/pdf.asp

```python
# b∫a  f(x)dx = P [ a < X ≤ b]
```

```python
#pdf for normal Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define parameters
mu = 0     # Mean
sigma = 1  # Standard deviation

# Generate x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Compute PDF
pdf_values = norm.pdf(x, mu, sigma)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, pdf_values, 'b-', label='Normal PDF')
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Normal Distribution PDF (μ=0, σ=1)")
plt.legend()
plt.grid()
plt.show()
```

```python
#pdf for normal Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define parameters
mu = 0     # Mean
sigma = 1  # Standard deviation

# Generate x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Compute PDF
pdf_values = norm.pdf(x, mu, sigma)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, pdf_values, 'b-', label='Normal PDF')
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Normal Distribution PDF (μ=0, σ=1)")
plt.legend()
plt.grid()
plt.show()
```

```python
# For Exponential Distribution
from scipy.stats import expon

lambda_ = 1.0  # Rate parameter
x = np.linspace(0, 5, 100)
pdf_values = expon.pdf(x, scale=1/lambda_)

plt.plot(x, pdf_values, 'r-', label='Exponential PDF')
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Exponential Distribution PDF (λ=1)")
plt.legend()
plt.grid()
plt.show()
```
