(f(x + h) - f(x)) / h

```python
import sympy as sp

# Define the symbol
x = sp.Symbol('x')

# Define the function
f = x**2 + 3*x + 2

# Compute the derivative
derivative = sp.diff(f, x)

# Show the result
print("Derivative of f(x):", derivative)
```

 Higher Order Derivatives

```python
# Second derivative
second_derivative = sp.diff(f, x, 2)
print("Second derivative of f(x):", second_derivative)
```

 Derivative of Trigonometric / Exponential Functions

```python
f = sp.sin(x) * sp.exp(x)
df = sp.diff(f, x)
print("Derivative:", df)
```
