```python
import sympy as sp

# Define variables
x, y = sp.symbols('x y')

# Define the vector-valued function
F = sp.Matrix([x**2 * y, sp.sin(x + y)])

# Define the variable vector
vars = sp.Matrix([x, y])

# Compute the Jacobian matrix
J = F.jacobian(vars)

print("Jacobian matrix:")
sp.pprint(J)
```
