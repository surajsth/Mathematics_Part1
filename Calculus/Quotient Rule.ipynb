* f(x)= U(x) / V(x)
* f ′(x)=  (u′(x) . v(x) − u(x)⋅v ′(x)) / v(x**2)


```python
import sympy as sp

# Define the symbol
x = sp.Symbol('x')
e = sp.E  # Euler's number

# Define the functions
u = x**2
v = e**x  # or use sp.exp(x) — both are valid

# Define the product rule function
def quotient_rule(u, v, var):
    numerator = sp.diff(u, var) * v - u * sp.diff(v, var)
    denominator = v**2
    return numerator / denominator

# Compute the derivative using the product rule
df_custom = quotient_rule(u, v, x)

# Display the result
print("Custom Quotient rule output:", df_custom)
```

```python
print("Simplified:", sp.simplify(df_custom))
```
