# Logarithmic Derivative

logarithmic derivative — a powerful trick, especially useful for simplifying the differentiation of products, quotients, or powers.The derivative of the natural logarithm function is the reciprocal of its
argument. It is frequently used in ML for optimization and logarithmic transformations. 0

```python
import sympy as sp

# Define the symbol
x = sp.Symbol('x')

# Define the function
f = x**2 * sp.sin(x)

# Take the natural log of f
ln_f = sp.log(f)

# Differentiate ln(f)
d_ln_f = sp.diff(ln_f, x)

# Multiply back by f(x) to get the derivative of f(x)
f_derivative = sp.simplify(f * d_ln_f)

# Print results
print("Logarithmic Derivative of f(x):", f_derivative)
```

```python
f = (x**2 + 1)**5 * sp.exp(x)
ln_f = sp.log(f)
d_ln_f = sp.diff(ln_f, x)
f_derivative = sp.simplify(f * d_ln_f)

print("Logarithmic Derivative:", f_derivative)
```

# Exponential Derivative

 exponential derivatives — a core part of calculus, especially useful in growth/decay models, differential equations, and of course... math wizardry
 The exponential function is unique as its derivative is equal to itself.
This property is key in gradient computations and exponential growth models in ML.

```python
import sympy as sp

x = sp.Symbol('x')
f = sp.exp(x)

df = sp.diff(f, x)
print("Derivative of e^x:", df)
```

```python
f = sp.exp(x**2)
df = sp.diff(f, x)
print("Derivative of e^(x^2):", df)
```
