```python
import sympy as sp

# Define the variable
x = sp.Symbol('x')

# Define a power function
f = x**5  # You can change the exponent

# Apply the derivative (Power Rule automatically handled by sympy)
df = sp.diff(f, x)

# Show result
print("Derivative of f(x):", df)
```

Custom Function to Apply Power Rule (Manually)

```python
def power_rule_derivative(coefficient, power):
    new_coefficient = coefficient * power
    new_power = power - 1
    return f"{new_coefficient} * x**{new_power}"

# Example: derivative of 3*x^4
print(power_rule_derivative(3, 4))  # Output: 12 * x**3
```
