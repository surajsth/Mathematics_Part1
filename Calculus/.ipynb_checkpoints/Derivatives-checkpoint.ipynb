```python
import sympy as sp

# Define the variables
x, y = sp.symbols('x y')

# Define the multivariable function
f = x**2 * y + y**3

# Partial derivatives
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

# Output
print("∂f/∂x =", df_dx)
print("∂f/∂y =", df_dy)
```

```python
# Second partial derivatives
d2f_dx2 = sp.diff(f, x, 2)   # ∂²f/∂x²
d2f_dy2 = sp.diff(f, y, 2)   # ∂²f/∂y²
d2f_dxdy = sp.diff(f, x, y)  # ∂²f/∂x∂y

print("∂²f/∂x² =", d2f_dx2)
print("∂²f/∂y² =", d2f_dy2)
print("∂²f/∂x∂y =", d2f_dxdy)
```

# Gradient


```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')

# Define the function
f = x**2 * y + y**3

# Compute the gradient
grad_f = [sp.diff(f, var) for var in (x, y)]

# Print result
print("Gradient ∇f =", grad_f)
```

```python
x, y, z = sp.symbols('x y z')
f = x**2 + y*z + sp.sin(x*z)

grad_f = [sp.diff(f, var) for var in (x, y, z)]
print("Gradient ∇f =", grad_f)
```

# Directional Derivative


```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')

# Define the function
f = x**2 * y + y**3

# Compute gradient
grad_f = [sp.diff(f, var) for var in (x, y)]

# Point at which to evaluate
point = {x: 1, y: 2}

# Evaluate gradient at the point
grad_at_point = [df.subs(point) for df in grad_f]

# Direction vector v = [3, 4], normalize it
import numpy as np
v = np.array([3, 4])
u = v / np.linalg.norm(v)  # unit vector

# Compute directional derivative: dot(grad_f, u)
directional_derivative = sum([float(grad_at_point[i]) * u[i] for i in range(2)])

print("Gradient at point:", grad_at_point)
print("Unit direction vector:", u)
print("Directional Derivative:", directional_derivative)
```

# Higher-Order Partial Derivatives


```python
import sympy as sp

# Define variables
x, y = sp.symbols('x y')

# Define the function
f = x**2 * y + y**3

# Compute second partial derivatives
f_xx = sp.diff(f, x, 2)     # ∂²f/∂x²
f_yy = sp.diff(f, y, 2)     # ∂²f/∂y²
f_xy = sp.diff(f, x, y)     # ∂²f/∂x∂y
f_yx = sp.diff(f, y, x)     # ∂²f/∂y∂x

# Display results
print("∂²f/∂x² =", f_xx)
print("∂²f/∂y² =", f_yy)
print("∂²f/∂x∂y =", f_xy)
print("∂²f/∂y∂x =", f_yx)
```

```python
f_xxy = sp.diff(f, x, 2, y)   # ∂³f/∂x²∂y
f_yyx = sp.diff(f, y, 2, x)   # ∂³f/∂y²∂x

print("∂³f/∂x²∂y =", f_xxy)
print("∂³f/∂y²∂x =", f_yyx)
```

# Total Derivative


```python
import sympy as sp

# Define variables
x, y, t = sp.symbols('x y t')

# Define the function f(x, y)
f = x**2 * y + y**3

# Compute partial derivatives
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

# Define x(t) and y(t)
x_t = t**2
y_t = sp.sin(t)

# Compute dx/dt and dy/dt
dx_dt = sp.diff(x_t, t)
dy_dt = sp.diff(y_t, t)

# Substitute x(t) and y(t) into df/dx and df/dy
df_dx_sub = df_dx.subs({x: x_t, y: y_t})
df_dy_sub = df_dy.subs({x: x_t, y: y_t})

# Compute total derivative df/dt
df_dt = df_dx_sub * dx_dt + df_dy_sub * dy_dt

# Simplify and print
df_dt_simplified = sp.simplify(df_dt)
print("Total derivative df/dt =", df_dt_simplified)
```

# Implicit Differentiation


```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')
y_func = sp.Function('y')(x)  # y is a function of x

# Define the implicit equation: x² + y² = 25
eq = x**2 + y_func**2 - 25

# Differentiate both sides with respect to x
d_eq = sp.diff(eq, x)

# Solve for dy/dx
dy_dx = sp.solve(d_eq, sp.Derivative(y_func, x))[0]

print("dy/dx =", dy_dx)
```
