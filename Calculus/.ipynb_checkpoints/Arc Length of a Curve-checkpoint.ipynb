```python
import sympy as sp

# Define symbols
x = sp.symbols('x')
y = sp.sin(x)

# Compute derivative dy/dx
dy_dx = sp.diff(y, x)

# Compute integrand √(1 + (dy/dx)²)
integrand = sp.sqrt(1 + dy_dx**2)

# Compute definite integral from 0 to π
arc_length = sp.integrate(integrand, (x, 0, sp.pi))

# Simplify or numerically evaluate
arc_length_approx = arc_length.evalf()

print("Exact Arc Length:", arc_length)
print("Approximate Arc Length:", arc_length_approx)
```
