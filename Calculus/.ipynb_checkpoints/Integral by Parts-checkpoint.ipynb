```python
import sympy as sp

# Define symbols
x = sp.symbols('x')

# Define the functions u and dv
u = x
dv = sp.exp(x)

# Compute du and v
du = sp.diff(u, x)
v = sp.integrate(dv, x)

# Apply the integration by parts formula
integral_by_parts = u * v - sp.integrate(v * du, x)

# Simplify the result
integral_by_parts_simplified = sp.simplify(integral_by_parts)
print("Integral by parts result:", integral_by_parts_simplified)
```

# Volume of Revolution (Disk Method)


```python
import sympy as sp

# Define the variable and function
x = sp.symbols('x')
f = sp.sqrt(x)

# Set the limits of integration (from 0 to 4)
a, b = 0, 4

# Apply the Disk Method formula: V = pi * integral(f(x)^2, x, a, b)
volume = sp.pi * sp.integrate(f**2, (x, a, b))

# Simplify and evaluate numerically
volume_simplified = sp.simplify(volume)
volume_approx = volume.evalf()

print("Exact volume:", volume_simplified)
print("Approximate volume:", volume_approx)
```

# Surface Integral

```python
import sympy as sp

# Define the variables
x, y = sp.symbols('x y')

# Define the scalar field f(x, y, z)
f = x**2 + y**2

# Parametrize the surface (upper hemisphere of a sphere)
z = sp.sqrt(4 - x**2 - y**2)

# Compute the partial derivatives of the parametric surface
dz_dx = sp.diff(z, x)
dz_dy = sp.diff(z, y)

# The surface element dS = sqrt(1 + (dz/dx)^2 + (dz/dy)^2) dx dy
dS = sp.sqrt(1 + dz_dx**2 + dz_dy**2)

# Surface integral: ∫∫ f(x, y) * dS over the domain of x, y
integral = sp.integrate(f * dS, (x, -2, 2), (y, -sp.sqrt(4 - x**2), sp.sqrt(4 - x**2)))

# Simplify the result
integral_simplified = sp.simplify(integral)
integral_approx = integral.evalf()

print("Exact surface integral result:", integral_simplified)
print("Approximate surface integral result:", integral_approx)
```

# Divergence of a Vector Field


```python
import sympy as sp

# Define the variables and the vector field components
x, y, z = sp.symbols('x y z')
F = sp.Matrix([x**2, y**2, z**2])

# Compute the divergence (∇ · F)
divergence_F = sp.diff(F[0], x) + sp.diff(F[1], y) + sp.diff(F[2], z)

# Simplify and display the result
divergence_F_simplified = sp.simplify(divergence_F)
print("Divergence of F(x, y, z) =", divergence_F_simplified)
```

# Curl of a Vector Field

```python
import sympy as sp

# Define the variables and the vector field components
x, y, z = sp.symbols('x y z')
F = sp.Matrix([x**2, y**2, z**2])

# Compute the curl (∇ × F)
curl_F = sp.Matrix([sp.diff(F[2], y) - sp.diff(F[1], z), 
                    sp.diff(F[0], z) - sp.diff(F[2], x), 
                    sp.diff(F[1], x) - sp.diff(F[0], y)])

# Simplify and display the result
curl_F_simplified = sp.simplify(curl_F)
print("Curl of F(x, y, z) =", curl_F_simplified)
```
