* The chain rule computes the derivative of a composite function. It is
extensively used in backpropagation for training neural networks.
(x)

```python
import sympy as sp

# Define the variable
x = sp.Symbol('x')

# Define the composite function f(x) = sin(x^2)
f = sp.sin(x**2)

# Take the derivative using sympy (applies chain rule automatically)
df = sp.diff(f, x)

# Print the result
print("Derivative using Chain Rule:", df)
```
