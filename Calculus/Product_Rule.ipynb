Example: For f(x) = (x**2)(e**x), f′(x) = 2xe**x + x**2e**x

```python
import sympy as sp

# Define the symbol
x = sp.Symbol('x')

# Define two functions
u = x**2
v = sp.sin(x)

# Apply the Product Rule
f = u * v
df = sp.diff(f, x)

print("Derivative of f(x) using Product Rule:", df)
```

```python
import sympy as sp

# Define the symbol
x = sp.Symbol('x')
e = sp.E  # Euler's number

# Define the functions
u = x**2
v = e**x  # or use sp.exp(x) — both are valid

# Define the product rule function
def product_rule(u, v, var):
    return sp.diff(u, var) * v + u * sp.diff(v, var)

# Compute the derivative using the product rule
df_custom = product_rule(u, v, x)

# Display the result
print("Custom product rule output:", df_custom)
```
