```python
import sympy as sp

# Define variable and function
x = sp.symbols('x')
f = sp.sin(x)

# First and second derivatives
f_prime = sp.diff(f, x)
f_double_prime = sp.diff(f_prime, x)

# Curvature formula
curvature = sp.Abs(f_double_prime) / (1 + f_prime**2)**(3/2)

# Simplify the curvature expression
curvature_simplified = sp.simplify(curvature)
print("Curvature κ(x) =", curvature_simplified)
```
